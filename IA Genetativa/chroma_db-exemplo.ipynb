```python
# importando as bibliotecas
import chromadb
import os
from chromadb.utils import embedding_functions

# Criando o cliente ChromaDB com armazenamento persistente
client = chromadb.PersistentClient(path="./chroma_data")

# Criando a coleção com a função de embedding
embedding_function = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)

# Criando ou obtendo a coleção
colecao = client.get_or_create_collection(
    name="base_conhecimento",
    embedding_function=embedding_function
)

# Adicionando documentos à coleção
docs = [
    "O contrato venceu em janeiro e precisa ser renovado.",
    "O sistema financeiro apresentou falha no pagamento.",
    "A política de segurança foi atualizada em março.",
    "O cliente solicitou cancelamento do serviço.",
    "A fatura foi paga com atraso."
]

# Gerando IDs para os documentos
ids = [f"doc-{i}" for i in range(len(docs))]

# Adicionando os documentos à coleção
colecao.add(documents=docs, ids=ids)

print("Indexado.")
```

```python
# Realizando uma consulta semântica
consulta = "documentos vencidos"
resultado = colecao.query(
    query_texts=[consulta],
    n_results=3
)

# Exibindo os resultados
print("\nResultados semânticos para:", consulta)
for doc in resultado["documents"][0]:
    print("-", doc)
```
