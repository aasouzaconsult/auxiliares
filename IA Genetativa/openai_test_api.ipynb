```python
from IPython.display import display, Markdown

from openai import OpenAI
client = OpenAI(api_key=openai.api_key)

#Monta a requisição para a API do GPT
def generate_chat_response(prompt):
    response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
      {
        "role": "user",
        "content": prompt
      }
    ],

    temperature=0.9,
    max_tokens=150)
    return response.choices[0].message.content

#Mostra resultados com aparência semelhante à interface web do ChatGPT
def display_chat(user_input, assistant_response):
    display(Markdown(f"**Usuário:** {user_input}"))
    display(Markdown(f"**Assistente:** {assistant_response}"))
```

### Interpretando uma piada 🤣

```python
prompt = """Qual a graça dessa piada:

Qual a diferença entre o lago e a padaria?
No lago, há sapinho. Na padaria, assa pão.

"""
result = generate_chat_response(prompt)

# Exibindo a conversa no estilo Usuárioa / Assistente
display_chat(prompt, result)
```

### Célula para interação ao modelo ChatGPT. Depois da célula iniciada, digite "exit" para sair.

```python
#Célula com While Loop para interação no modelo Chat. Digite "exit" para sair.
while True:
  prompt = input("Faça uma pergunta ao ChatGPT: \n")
  if prompt.lower() == "exit":
    break
  else:
    pass
  result = generate_chat_response(prompt)

  # Exibindo a conversa no estilo do site da OpenAI
  display_chat(prompt, result)
```

### Concatenando resultados e testando roles (contexto)
Geração de sequencia de prompts...

```python
def generate_chat_response1(prompt):
    response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {"role": "system", "content": "Atue como especialista em dados. Respostas sempre diretas e objetivas"},
        {"role": "assistant", "content": "Dados é uma area facinante e que tá em alta no mercado, usa muito SQL no seu dia a dia, Python, Modelagem de Dados, Visualização, AED"},
        {"role": "user",  "content": prompt}
    ],
    temperature=0.9,
    max_tokens=1000)
    return response.choices[0].message.content
```

```python
import re

# pode ser utilizada tb o https://pypi.org/project/tiktoken/
def contar_tokens(texto):
    # Separando palavras e caracteres especiais
    tokens = re.findall(r'\w+|[^\w\s]', texto, re.UNICODE)
    return len(tokens)

# String para armazenar o histórico de prompts e respostas
historico_respostas = ""
contagem_tokens = 0

# Célula com While Loop para interação no modelo Chat. Digite "exit" para sair.
while True:
    prompt = input("Faça uma pergunta ao ChatGPT: \n")
    if prompt.lower() == "exit":
        break

    # Atualizando o histórico e a contagem de tokens
    historico_respostas += "Pergunta: " + prompt + "\n"
    contagem_tokens += contar_tokens(prompt)

    # Chama a função de geração de resposta com o histórico completo
    result = generate_chat_response1(historico_respostas)

    historico_respostas += "Resposta: " + result + "\n"
    contagem_tokens += contar_tokens(result)

    print(f"Total de tokens até agora: {contagem_tokens}")

    # Exibindo a conversa no estilo do site da OpenAI
    display_chat(prompt, result)
```
